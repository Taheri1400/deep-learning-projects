از هرگونه پیش‌پردازشی که فکر می‌کنید به بهبود عملکرد مدل و فرآیند آموزش آن کمک می‌کند بهره ببرید. از جمله کارهایی که شاید در این مرحله به آن نیاز پیدا کنید می‌توان به تغییر اندازه‌ی تصاویر، نرمال‌سازی مقادیر تصاویر، استفاده از تکنیک‌های تقویت داده (داده‌افزایی)، اعمال تابع پیش‌پردازش یک مدل پیش‌آموخته‌ی مشخص و غیره اشاره کرد.

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

```python
data_augmentation = keras.Sequential([
    layers.RandomFlip('horizontal'),
    layers.RandomRotation(0.05),
    layers.RandomZoom(0.1)
])

# ResNet50 preprocessing
preprocess_input = keras.applications.resnet50.preprocess_input

# Apply preprocessing to training data
train_dataset = train_dataset.map(
    lambda images, labels: (
        preprocess_input(data_augmentation(images, training=True)),
        labels
    ),
    num_parallel_calls=tf.data.AUTOTUNE
)

# Apply preprocessing to validation data
valid_dataset = valid_dataset.map(
    lambda images, labels: (
        preprocess_input(images),
        labels
    ),
    num_parallel_calls=tf.data.AUTOTUNE
)

# Apply preprocessing to test data
test_dataset = test_dataset.map(
    lambda images: preprocess_input(images),
    num_parallel_calls=tf.data.AUTOTUNE
)

# Improve performance
train_dataset = train_dataset.prefetch(tf.data.AUTOTUNE)
valid_dataset = valid_dataset.prefetch(tf.data.AUTOTUNE)
test_dataset = test_dataset.prefetch(tf.data.AUTOTUNE)
```

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazir" size=3>
در این پروژه کاملاً دست‌تان باز است تا ساختارهای دلخواه خود را پیاده و آزمایش کنید. می‌توانید یک شبکه‌ی عصبی کانولوشنی طراحی کنید و از تکنیک‌های رایج این نوع شبکه‌ها بهره ببرید تا به مدلی با عملکرد مناسب دست یابید، یا می‌توانید از یک شبکه‌ی پیش‌آموخته کمک گرفته و تصمیم بگیرید که چه لایه‌هایی را به آن اضافه کنید یا چه بخش‌هایی از آن را فریز کرده یا آموزش دهید. همچنین استفاده از تکنیک‌های دیگری که در فصل بهینه‌سازی و هایپرپارامتر آموخته‌اید را فراموش نکنید.
</font>

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Load pretrained ResNet50
base_model = keras.applications.ResNet50(
    weights='imagenet',
    include_top=False,
    input_shape=(160, 160, 3),
    pooling='avg'
)

# Freeze ResNet50 layers
base_model.trainable = True

for layer in base_model.layers[:-30]:
    layer.trainable = False

# Build classification head
inputs = keras.Input(shape=(160, 160, 3))

x = base_model(inputs, training=False)

x = layers.Dense(512, activation='relu')(x)
x = layers.BatchNormalization()(x)
x = layers.Dropout(0.5)(x)

x = layers.Dense(256, activation='relu')(x)
x = layers.BatchNormalization()(x)
x = layers.Dropout(0.4)(x)

x = layers.Dense(128, activation='relu')(x)
x = layers.Dropout(0.3)(x)

outputs = layers.Dense(11, activation='softmax')(x)

model = keras.Model(inputs, outputs)
```

```python
# Compile model
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Early stopping
early_stopping = keras.callbacks.EarlyStopping(
    monitor='val_accuracy',
    patience=4,
    restore_best_weights=True
)

# Reduce learning rate when validation accuracy stops improving
reduce_lr = keras.callbacks.ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.2,
    patience=2,
    min_lr=1e-7
)
```

```python
history = model.fit(
    train_dataset,
    validation_data=valid_dataset,
    epochs=10,
    callbacks=[early_stopping, reduce_lr]
)
```

```python
loss, accuracy = model.evaluate(valid_dataset)

print(f"Validation Accuracy: {accuracy * 100:.2f}%")
```

```python
import os
import tensorflow as tf
import keras
import pandas as pd
import numpy as np

# 1. بارگذاری تصاویر آزمون
test_dataset = keras.utils.image_dataset_from_directory(
    './test',
    image_size=(160, 160),
    batch_size=32,
    labels=None,
    shuffle=False
)

# 2. استخراج نام فایل‌ها با همان ترتیب دیتاست
image_names = [
    os.path.basename(path)
    for path in test_dataset.file_paths
]

# 3. پیش‌پردازش تصاویر برای ResNet50
preprocess_input = keras.applications.resnet50.preprocess_input

test_dataset = test_dataset.map(
    lambda images: preprocess_input(images),
    num_parallel_calls=tf.data.AUTOTUNE
)

# 4. پیش‌بینی
predictions = model.predict(test_dataset)

# 5. تبدیل خروجی مدل به شماره کلاس
predicted_indices = np.argmax(predictions, axis=1)

# ترتیب واقعی نام کلاس‌ها
class_names = test_dataset.class_names if hasattr(test_dataset, 'class_names') else [
    '0', '1', '10', '2', '3', '4', '5', '6', '7', '8', '9'
]

cat_ids = [int(class_names[i]) for i in predicted_indices]

# 6. ساخت و ذخیره فایل خروجی
submission = pd.DataFrame({
    'image_name': image_names,
    'cat_id': cat_ids
})

submission.to_csv('submission.csv', index=False)

print(submission.head())
print("Submission shape:", submission.shape)
```

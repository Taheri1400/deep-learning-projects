<h2 dir=rtl align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazir" color="#0099cc">
رویکرد <code>"int"</code>

```python
docs = ['یک جام شراب صد دل و دین ارزد',
        'یک جرعهٔ می مملکت چین ارزد']
```

```python
vectorizer = keras.layers.TextVectorization()
vectorizer.adapt(docs)
```

```python
vocab = vectorizer.get_vocabulary()
vocab_size = vectorizer.vocabulary_size()
for i in range(0, vocab_size):
  print(i, vocab[i])
```

```python
docs = ['یک جام شراب صد دل و دین ارزد',
        'یک جرعهٔ می مملکت چین ارزد']

print(vectorizer(docs))
```

```python
new_docs = ['جز بادهٔ لعل نیست در روی زمین',
            'تلخی که هزار جان شیرین ارزد']

vectorizer(new_docs)
```

<h2 dir=rtl align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazir" color="#0099cc">
رویکرد <code>"multi_hot"</code>

```python
vectorizer = keras.layers.TextVectorization(output_mode='multi_hot')
vectorizer.adapt(docs)

vocab = vectorizer.get_vocabulary()
vocab_size = vectorizer.vocabulary_size()
for i in range(0, vocab_size):
  print(i, vocab[i])
```

```python
print(vectorizer(docs))
```

<h2 dir=rtl align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazir" color="#0099cc">
رویکرد <code>"count"</code>
</font>
</h2>

```python
docs = ['یک جام شراب صد دل و دین ارزد یک جرعهٔ می مملکت چین ارزد',
        'جز بادهٔ لعل نیست در روی زمین تلخی که هزار جان شیرین ارزد']

vectorizer = keras.layers.TextVectorization(output_mode='count')
vectorizer.adapt(docs)

print(vectorizer(docs))
```

<h2 dir=rtl align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazir" color="#0099cc">
رویکرد <code>"tf_idf"</code>
</font>
</h2>

```python
docs = ['یک جام شراب صد دل و دین ارزد یک جرعهٔ می مملکت چین ارزد',
        'جز بادهٔ لعل نیست در روی زمین تلخی که هزار جان شیرین ارزد']

vectorizer = keras.layers.TextVectorization(output_mode='tf_idf')
vectorizer.adapt(docs)

print(vectorizer(docs))
```

```python
vocab = vectorizer.get_vocabulary()
vocab_size = vectorizer.vocabulary_size()
for i in range(0, vocab_size):
  print(i, vocab[i])
```

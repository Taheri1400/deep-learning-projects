```python
import tensorflow as tf
from tensorflow import keras
```

<h2 dir=rtl align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazir" color="#0099cc">
ساخت شبکه به‌صورت سطح پایین
</font>
</h2>

```python
class QDense(tf.Module):
    def __init__(self, in_features, out_features, activation=tf.identity, name=None):
        super().__init__(name=name)

        # Xavier/Glorot initialization for weights
        initializer = tf.initializers.GlorotUniform()
        self.w = tf.Variable(initializer(shape=[in_features, out_features]), name='w')

        self.b = tf.Variable(tf.zeros([out_features]), name='b')

        self.activation = activation

    def __call__(self, x):
        y = tf.matmul(x, self.w) + self.b
        y = self.activation(y)
        return y
```

```python
class QSequentialModel(tf.Module):
  def __init__(self, name=None):
    super().__init__(name=name)

    self.dense_1 = QDense(in_features=784, out_features=20, activation=tf.nn.relu)
    self.dense_2 = QDense(in_features=20, out_features=10, activation=tf.nn.softmax)

  def __call__(self, x):
    x = self.dense_1(x)
    return self.dense_2(x)
```

```python
model = QSequentialModel(name="Quera")
```

استفاده از <code>tf.keras.Model</code> به‌جای <code>tf.Module</code>

```python
class QDense(tf.keras.Model):
    def __init__(self, in_features, out_features, activation=tf.identity, name=None):
        super(QDense, self).__init__(name=name)

        # Xavier/Glorot initialization for weights
        initializer = tf.initializers.GlorotUniform()
        self.w = tf.Variable(initializer(shape=[in_features, out_features]), name='w')

        self.b = tf.Variable(tf.zeros([out_features]), name='b')

        self.activation = activation

    def call(self, x):
        y = tf.matmul(x, self.w) + self.b
        y = self.activation(y)
        return y
```

```python
class QSequentialModel(tf.keras.Model):
  def __init__(self, name=None):
    super(QSequentialModel, self).__init__(name=name)

    self.dense_1 = QDense(in_features=784, out_features=20, activation=tf.nn.relu)
    self.dense_2 = QDense(in_features=20, out_features=10, activation=tf.nn.softmax)

  def call(self, x):
    x = self.dense_1(x)
    return self.dense_2(x)
```

```python
model = QSequentialModel(name="Quera")
```

```python
def xavier_init(shape):
  # Computes the xavier initialization values for a weight matrix
  in_dim, out_dim = shape
  xavier_lim = tf.sqrt(6.)/tf.sqrt(tf.cast(in_dim + out_dim, tf.float32))
  weight_vals = tf.random.uniform(shape=(in_dim, out_dim),
                                  minval=-xavier_lim, maxval=xavier_lim, seed=22)
  return weight_vals
```

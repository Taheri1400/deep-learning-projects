```python
docs = ['یک جام شراب صد دل و دین ارزد',
        'یک جرعهٔ می مملکت چین ارزد']

vectorizer = keras.layers.TextVectorization()
vectorizer.adapt(docs)

VOCAB_SIZE = vectorizer.vocabulary_size()
print("Vocab size:", VOCAB_SIZE)
```

```python
BATCH_SIZE = 32
WINDOW_LEN = 8
EMB_SIZE = 64

model = keras.Sequential()
model.add(keras.layers.Input(batch_shape=(BATCH_SIZE, WINDOW_LEN), dtype='int32'))
model.add(keras.layers.Embedding(input_dim=VOCAB_SIZE, output_dim=EMB_SIZE))
model.add(keras.layers.LSTM(128))
model.add(keras.layers.Dense(10))
model.summary()
```

```python
from dadmatools.embeddings import get_embedding, get_all_embeddings_info, get_embedding_info

# get embedding information
embedding_info = get_embedding_info('glove-wiki')

# load embedding
word_embedding = get_embedding('glove-wiki')

# get vector of the word
print(word_embedding['جام'])
print(len(word_embedding['جام']))
```

```python
import numpy as np

embedding_dimension = 50
vocab = vectorizer.get_vocabulary()

embedding_matrix = np.zeros((len(vocab), embedding_dimension))

for idx, word in enumerate(vocab):
  try:
    embedding_vector = word_embedding[word]
    embedding_matrix[idx] = embedding_vector[:embedding_dimension]
  except:
    pass
```

```python
BATCH_SIZE = 32
WINDOW_LEN = 8

model = keras.Sequential()
model.add(keras.layers.Input(batch_shape=(BATCH_SIZE, WINDOW_LEN), dtype='int32'))
model.add(keras.layers.Embedding(input_dim=embedding_matrix.shape[0],
                                 output_dim=embedding_matrix.shape[1],
                                 name='embedding'))
model.add(keras.layers.LSTM(128))
model.add(keras.layers.Dense(10))
model.summary()
```

```python
model.get_layer('embedding').set_weights([embedding_matrix])
```

```python
model.get_layer('embedding').trainable = False
```

```python
from sentence_transformers import SentenceTransformer
from sentence_transformers.util import cos_sim
import numpy as np

# Loading model BGE-M3
model = SentenceTransformer('BAAI/bge-m3')

# A collection of Persian sentences for testing
texts = [
    "سلام، مدل BGE-M3 چیست؟",
    "این یک مدل تعبیه‌سازی متن چندزبانه است که از زبان فارسی هم پشتیبانی می‌کند.",
    "هوا امروز خیلی عالیه برای بیرون رفتن.",
    "امروز هوا برای گردش در طبیعت مناسب است.",
    "یادگیری ماشین یک حوزه جذاب در علوم کامپیوتر است.",
    "هوش مصنوعی و یادگیری عمیق از مباحث مهم در فناوری اطلاعات هستند.",
    "کتاب خواندن یکی از بهترین راه‌ها برای افزایش دانش است.",
    "مطالعه کردن به گسترش افق دید کمک می‌کند.",
    "تهران پایتخت ایران و یک شهر بزرگ و پرجمعیت است.",
    "اصفهان شهری تاریخی با معماری زیباست.",
    "امروز می‌خواهم در مورد موضوع جدیدی تحقیق کنم.",
    "بررسی و پژوهش در زمینه‌های مختلف علمی اهمیت دارد.",
    "چالش‌های پیش روی هوش مصنوعی در آینده چیست؟",
    "مسائل و مشکلات آینده در حوزه هوش مصنوعی.",
    "آیا مدل‌های زبانی بزرگ می‌توانند شعر بگویند؟",
    "توانایی مدل‌های LLM در سرودن شعر.",
    "بهترین روش برای یادگیری زبان انگلیسی چیست؟",
    "مؤثرترین شیوه‌های آموزش زبان انگلیسی.",
    "فواید مدیتیشن برای سلامت روان.",
    "تأثیرات مثبت مراقبه بر آرامش ذهنی.",
    "تاریخچه زبان فارسی به قرون گذشته برمی‌گردد.",
    "ریشه‌های زبان پارسی در اعماق تاریخ قرار دارد."
]

# Generating embedding vectors
embeddings = model.encode(texts, convert_to_tensor=True, show_progress_bar=True)
```

```python
# Sentence pairs for comparison
pairs_to_compare = [
    (0, 1),  # درباره مدل BGE-M3
    (2, 3),  # درباره هوا
    (4, 5),  # درباره یادگیری ماشین و هوش مصنوعی
    (6, 7),  # درباره کتاب خواندن و مطالعه
    (8, 9),  # درباره تهران و اصفهان
    (10, 11),  # درباره تحقیق و پژوهش
    (12, 13),  # درباره چالش‌های هوش مصنوعی
    (14, 15),  # درباره توانایی مدل‌های زبانی
    (16, 17),  # درباره یادگیری زبان انگلیسی
    (18, 19),  # درباره مدیتیشن
    (20, 21)   # درباره تاریخچه زبان فارسی
]

# Calculating and displaying cosine similarity
print("نتایج شباهت معنایی بین جفت‌های جملات:")
for i, j in pairs_to_compare:
    similarity = cos_sim(embeddings[i], embeddings[j]).item()
    print(f"شباهت بین '{texts[i]}' و '{texts[j]}': {similarity:.4f}")
```

```python
from FlagEmbedding import BGEM3FlagModel

# Loading the model with FlagEmbedding
flag_model = BGEM3FlagModel('BAAI/bge-m3', use_fp16=True)

# Generating sparse embeddings for a pair of sentences
sentence_pair = [texts[0], texts[1]]  # Example: About model BGE-M3
sparse_emb = flag_model.encode(sentence_pair, return_sparse=True)['lexical_weights']

# Displaying lexical weights
print("وزن‌های لکسیکال برای جفت جملات:")
for i, sentence in enumerate(sentence_pair):
    print(f"\nوزن‌های لکسیکال برای '{sentence}':")
    token_weights = sparse_emb[i]
    # Iterate over the token IDs (keys) and weights (values)
    for token_id_str, weight in token_weights.items():
        try:
            token_id = int(token_id_str) # Attempt to convert to integer
            # Use the tokenizer's decode method to get the token string
            token = flag_model.tokenizer.decode([token_id])
            print(f"  {token}: {weight:.4f}")
        except ValueError:
            print(f"Could not convert token_id '{token_id_str}' to integer. Skipping.")
```
